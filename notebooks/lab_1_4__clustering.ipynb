# Практикум №4: Кластеризація даних різної форми алгоритмами навчання без вчителя. Оцінка якості результатів кластеризації

0. Імпорт залежностей.

```python
import random
import numpy as np

from sklearn.datasets import make_blobs
from sklearn.cluster import MeanShift
from sklearn.metrics import (
    adjusted_rand_score,
    adjusted_mutual_info_score,
    silhouette_score,
)

from src.visualization.plot import plot_2d_data, plot_mean_shift_cluster_score

random_seed = 42

random.seed(random_seed)
np.random.seed(random_seed)
rng = np.random.RandomState(random_seed)
```

1. Представити початковi данi графiчно.

```python
X1, y_true = make_blobs(n_samples=40000, centers=4, cluster_std=0.60, random_state=0)

x = np.dot(X1, rng.randn(2, 2))
```

```python
plot_2d_data(x, y_true)
```

2. Побудувати модель кластеризацiї згiдно з варiантом.

```python
model = MeanShift()
```

3. Виконати кластеризацiю даних на основi моделi.

```python
clustering = model.fit(x)
```

4. Представити розбиття на кластери графiчно, наприклад, рiзними кольорами.

```python
plot_2d_data(x, clustering.labels_)
```

5. Розрахувати час кластеризацiї. Оцiнити швидкодiю методу на надвеликих наборах даних (наприклад, при збiльшеннi кiлькостi точок даних до ста тисяч i бiльше).

```python
big_X1, big_y_true = make_blobs(
    n_samples=100000, centers=4, cluster_std=0.60, random_state=0
)

big_x = np.dot(big_X1, rng.randn(2, 2))
```

```python
plot_2d_data(big_x, big_y_true)
```

```python
clustering = model.fit(big_x)
```

```python
plot_2d_data(big_x, clustering.labels_)
```

6. Побудувати кiлька альтернативних моделей:
* шляхом змiни значень параметрiв основної моделi,
* використати рiзнi функцiї вiдстанi в алгоритмах, де це можна зробити,
* задати рiзнi значення кiлькостi кластерiв, в алгоритмах де кiлькiсть кластерiв - параметр.

```python
estimated_number_of_clusters_arr = []
adjusted_rand_index_arr = []
adjusted_mutual_information_arr = []
silhouette_coefficient_arr = []
bandwidth_arr = np.arange(0.1, 2.0, 0.1)

for it in bandwidth_arr:
    print("START: bandwidth=" + str(it))
    model = MeanShift(bandwidth=it)
    clustering = model.fit(x)
    y_pred = clustering.labels_
    plot_2d_data(x, y_pred)

    estimated_number_of_clusters = len(np.unique(y_pred))
    adjusted_rand_index = adjusted_rand_score(y_true, y_pred)
    adjusted_mutual_information = adjusted_mutual_info_score(y_true, y_pred)
    silhouette_coefficient = silhouette_score(x, y_pred)

    estimated_number_of_clusters_arr.append(estimated_number_of_clusters)
    adjusted_rand_index_arr.append(adjusted_rand_index)
    adjusted_mutual_information_arr.append(adjusted_mutual_information)
    silhouette_coefficient_arr.append(silhouette_coefficient)

    print(
        f"END: estimated_number_of_clusters={estimated_number_of_clusters}, adjusted_rand_index={adjusted_rand_index}, adjusted_mutual_information={adjusted_mutual_information}, silhouette_coefficient={silhouette_coefficient}"
    )
```

7. Для кожної альтернативної моделi розрахувати метрики якостi кластеризацiї, що реалiзованi в модулi sklearn.metrics (тiльки метрики згiдно з варiантом):
* Estimated Number of Clusters.
* Adjusted Rand Index.
* Adjusted Mutual Information.
* Homogeneity.
* Completeness.
* V-measure.
* Silhouette Coefficient.
* Calinski-Harabasz Index.
* Davies-Bouldin index.
* Contingency Matrix.

```python
plot_mean_shift_cluster_score(
    bandwidth_arr, estimated_number_of_clusters_arr, "Estimated number of clusters"
)
plot_mean_shift_cluster_score(
    bandwidth_arr, adjusted_rand_index_arr, "Adjusted Rand Index"
)
plot_mean_shift_cluster_score(
    bandwidth_arr, adjusted_mutual_information_arr, "Adjusted Mutual Information"
)
plot_mean_shift_cluster_score(
    bandwidth_arr, silhouette_coefficient_arr, "Silhouette Coefficient"
)
```
