```python
import random
import numpy as np
import pandas as pd

from mlxtend.frequent_patterns import apriori, fpgrowth
from mlxtend.preprocessing import TransactionEncoder

from src.data.util.dataset import download_and_unzip
from src.definitions import EXTERNAL_DATA_FOLDER

random_seed = 42

random.seed(random_seed)
np.random.seed(random_seed)
rng = np.random.RandomState(random_seed)
```

1. Взяти файл з даними у вiдповiдностi з варiантом.

```python
ds_path = download_and_unzip(
    "sulmansarwar", "transactions-from-a-bakery", EXTERNAL_DATA_FOLDER
)
df = pd.read_csv(ds_path / "BreadBasket_DMS.csv")

df.head()
```

```python
df = df.drop(["Date", "Time"], axis=1)
df = df.dropna()
```

```python
grouped = df.groupby("Transaction")["Item"].apply(set).apply(list).tolist()
```

```python
te = TransactionEncoder()
te_ary = te.fit(grouped).transform(grouped)
df = pd.DataFrame(te_ary, columns=te.columns_)
df
```

2. Побудувати частi набори та асоцiативнi правила (АП), використовуючи алгоритм Apriori. Дослiдити множини АП залежно вiд параметрiв побудови АП:
* мiнiмальної пiдтримки,
* мiнiмальної достовiрностi,
* максимальної потужностi частих наборiв.<br/><br/>
Пiдiбрати значення параметрiв алгоритму Apriori, якi призводять до значущих АП. АП Rk вважати значущим, якщо покращення (improvement або лiфт (lift)): Improv(Rk) > 1.

```python
apriori(df, min_support=0.05)
```

3. Побудувати частi набори та множину АП, використовуючи алгоритм FP-росту. Дослiдити множини АП залежно вiд параметрiв побудови АП, якi було наведено вище для Apriori. Пiдiбрати значення параметрiв алгоритму FP-росту, якi призводять до значущих АП.
